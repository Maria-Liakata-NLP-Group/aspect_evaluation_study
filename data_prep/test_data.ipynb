```python
import pandas as pd
```

```python
# import csv file as dataframe
test_data = pd.read_csv('New datasets subset C abductive - vitc_john.csv')
# only select first 25 rows
test_data = test_data.head(25)

# only select columns with the following names: claim, evidence, label
test_data = test_data[['claim', 'evidence', 'label']]

# rename label to veracity
test_data = test_data.rename(columns={'label': 'veracity'})

# give each row a unique id
test_data['id'] = [f'test_{i+1}' for i in range(len(test_data))]

# give each row a batch number
test_data['batch'] = 'test_1'

# save dataframe as json file
test_data.to_json('test_data.json', orient='records')
```
